# Clinical NLP Pipeline Demo

Demonstrates ClinicalBERT NER pipeline and ICU risk forecasting on healthcare data.

## 1. Setup


```python
from src.nlp.clinical_ner import ClinicalNERPipeline
from src.models.icu_risk_model import ICURiskModel
print('Clinical AI modules loaded')
```

## 2. Clinical NER - Entity Extraction


```python
pipeline = ClinicalNERPipeline(
    bert_model='emilyalsentzer/Bio_ClinicalBERT',
    spacy_model='en_core_sci_lg'
)

note = """
Patient is a 65-year-old male with history of hypertension and diabetes mellitus.
Presenting with shortness of breath and chest pain. Blood pressure 180/110.
Started on metformin 500mg twice daily. Denies fever or chills.
"""

result = pipeline.process_note('NOTE_001', 'PAT_001', note)

print(f'Diagnoses: {result.diagnoses}')
print(f'Medications: {result.medications}')
print(f'Negated Conditions: {result.negated_conditions}')
print(f'Total Entities: {len(result.entities)}')
```

## 3. ICU Risk Prediction


```python
import pandas as pd
import numpy as np

# Sample patient vitals and labs
patient_data = pd.DataFrame([{
    'heart_rate_mean': 92, 'heart_rate_max': 118, 'sbp_mean': 105,
    'dbp_mean': 68, 'respiratory_rate_mean': 22, 'spo2_min': 94,
    'temperature_max': 38.6, 'sofa_score': 8, 'apache_ii_score': 21,
    'gcs_min': 13, 'age': 65, 'is_male': 1, 'bmi': 28.5,
    'lactate_max': 3.2, 'creatinine_max': 1.8, 'wbc_max': 14.2,
    'has_diabetes': 1, 'has_chf': 0, 'has_copd': 0, 'has_ckd': 0, 'has_cancer': 0,
    'icu_los_days': 2.5, 'ventilated_flag': 0, 'vasopressor_flag': 1,
    'num_prior_admissions': 2, 'hours_since_admission': 12,
    'platelet_min': 145, 'hemoglobin_min': 10.2, 'sodium_min': 136,
    'potassium_max': 4.1, 'bilirubin_max': 1.1,
}])

model = ICURiskModel.load('models/icu_risk')
predictions = model.predict_mortality_risk(patient_data)
print(f'Mortality Probability: {predictions["mortality_probability"].values[0]:.1%}')
print(f'Risk Tier: {predictions["risk_tier"].values[0]}')
```
